```python
clicks=df[['ORGANISATION_ID', 'ORGANISATION_VERTICAL',
       'ORGANISATION_SUBVERTICAL', 'ORGANISATION_MARKETING_SOURCES',
       'ORGANISATION_PRIMARY_TERRITORY_NAME', 'TERRITORY_NAME', 'DATE_DAY',
       'CURRENCY_CODE','GOOGLE_PAID_SEARCH_SPEND',
       'GOOGLE_SHOPPING_SPEND', 'GOOGLE_PMAX_SPEND', 'GOOGLE_DISPLAY_SPEND',
       'GOOGLE_VIDEO_SPEND', 'GOOGLE_PAID_SEARCH_CLICKS',
       'GOOGLE_SHOPPING_CLICKS', 'GOOGLE_PMAX_CLICKS', 'GOOGLE_DISPLAY_CLICKS',
       'GOOGLE_VIDEO_CLICKS','Total_spends',
       'Total_spend_GBP', 'Total_Revenue', 'Total_Revenue_GBP']]
clicks
```

```python
clicks['Google_total_spend']=clicks.loc[:,['GOOGLE_PAID_SEARCH_SPEND','GOOGLE_SHOPPING_SPEND','GOOGLE_PMAX_SPEND','GOOGLE_DISPLAY_SPEND','GOOGLE_VIDEO_SPEND']].sum(axis=1)
```

```python
temp_list3=clicks.loc[
    (clicks.GOOGLE_PAID_SEARCH_CLICKS.isnull()) &
    (clicks.GOOGLE_SHOPPING_CLICKS.isnull()) &
    (clicks.GOOGLE_PMAX_CLICKS.isnull()) &
    (clicks.GOOGLE_DISPLAY_CLICKS.isnull()) &
    (clicks.GOOGLE_VIDEO_CLICKS.isnull())].index.tolist()
clicks=clicks.drop(temp_list3)
```

```python
temp_list5=clicks.loc[
    (clicks.GOOGLE_PAID_SEARCH_SPEND.isnull()) &
    (clicks.GOOGLE_SHOPPING_SPEND.isnull()) &
    (clicks.GOOGLE_PMAX_SPEND.isnull()) &
    (clicks.GOOGLE_DISPLAY_SPEND.isnull()) &
    (clicks.GOOGLE_VIDEO_SPEND.isnull())
].index.tolist()
clicks=clicks.drop(temp_list5)
```

```python
clicks.GOOGLE_PAID_SEARCH_SPEND=clicks.GOOGLE_PAID_SEARCH_SPEND.fillna(0)
clicks.GOOGLE_SHOPPING_SPEND=clicks.GOOGLE_SHOPPING_SPEND.fillna(0)
clicks.GOOGLE_PMAX_SPEND=clicks.GOOGLE_PMAX_SPEND.fillna(0)
clicks.GOOGLE_DISPLAY_SPEND=clicks.GOOGLE_DISPLAY_SPEND.fillna(0)
clicks.GOOGLE_VIDEO_SPEND=clicks.GOOGLE_VIDEO_SPEND.fillna(0)   
```

```python
clicks.GOOGLE_PAID_SEARCH_CLICKS=clicks.GOOGLE_PAID_SEARCH_CLICKS.fillna(0)
clicks.GOOGLE_SHOPPING_CLICKS=clicks.GOOGLE_SHOPPING_CLICKS.fillna(0)
clicks.GOOGLE_PMAX_CLICKS=clicks.GOOGLE_PMAX_CLICKS.fillna(0)
clicks.GOOGLE_DISPLAY_CLICKS=clicks.GOOGLE_DISPLAY_CLICKS.fillna(0)
clicks.GOOGLE_VIDEO_CLICKS=clicks.GOOGLE_VIDEO_CLICKS.fillna(0)
```

```python
clicks['ORGANISATION_VERTICAL'].isnull().sum()
```

```python
clicks['Google_total_clicks']=clicks.loc[:,['GOOGLE_PAID_SEARCH_CLICKS','GOOGLE_SHOPPING_CLICKS','GOOGLE_PMAX_CLICKS','GOOGLE_DISPLAY_CLICKS',
       'GOOGLE_VIDEO_CLICKS']].sum(axis=1)
```

```python
clicks['Google_cost_perclick']=clicks.Google_total_clicks/clicks.Google_total_spend
```

```python
clicks
```

```python
clicks.isnull().sum()
```

```python
clicks.drop(clicks.loc[clicks.Google_cost_perclick.isnull()].index.tolist(), inplace=True)
```

```python
clicks.drop(clicks.loc[
    (clicks.Google_total_spend == 0)
    |
    (clicks.Google_cost_perclick == 0)
].index.tolist(), inplace = True)
```

```python
spend_vs_costperclick = clicks.groupby('ORGANISATION_VERTICAL')[['Google_total_spend', 'Google_cost_perclick']].mean().reset_index()
display(spend_vs_costperclick)

fig = ob.Figure()

fig.add_trace(
    ob.Bar(
        x = spend_vs_costperclick['ORGANISATION_VERTICAL'],
        y = spend_vs_costperclick['Google_total_spend'],
        yaxis = 'y'
    )
)

fig.add_trace(
    ob.Scatter(
        x = spend_vs_costperclick['ORGANISATION_VERTICAL'],
        y = spend_vs_costperclick['Google_cost_perclick'],
        yaxis = 'y2'
    )
)

fig.update_layout(
    yaxis2 = dict(
        anchor="y",
        overlaying="y",
        side="right",
        # position=0.15
    )
)

fig.show()
```

## Cross-Channel Interaction Analysis:
### Correlation between channels

```python
channels = [
'GOOGLE_PAID_SEARCH_SPEND',
'GOOGLE_SHOPPING_SPEND',
'GOOGLE_PMAX_SPEND',
'GOOGLE_DISPLAY_SPEND',
'GOOGLE_VIDEO_SPEND',
'META_FACEBOOK_SPEND',
'META_INSTAGRAM_SPEND',
'META_OTHER_SPEND',
'TIKTOK_SPEND'
]

correlation_matrix = df[channels].corr()

print(correlation_matrix)
```

### Halo effects identification between paid and organic channels

```python
df['TOTAL_PAID_SPEND'] = (
    df['GOOGLE_PAID_SEARCH_SPEND'] +
    df['GOOGLE_SHOPPING_SPEND'] +
    df['GOOGLE_PMAX_SPEND'] +
    df['META_FACEBOOK_SPEND'] +
    df['META_INSTAGRAM_SPEND'] +
    df['TIKTOK_SPEND']
)

# correlation between paid and organic
halo_corr = df[['TOTAL_PAID_SPEND','FIRST_PURCHASES']].corr()

print(halo_corr)
```

### Positive correlation means possible halo effect

### Cannibalization analysis between similar channels

```python
cannibalization = [
'GOOGLE_PAID_SEARCH_SPEND',
       'GOOGLE_SHOPPING_SPEND', 'GOOGLE_PMAX_SPEND', 'GOOGLE_DISPLAY_SPEND',
       'GOOGLE_VIDEO_SPEND', 'META_FACEBOOK_SPEND', 'META_INSTAGRAM_SPEND',
       'META_OTHER_SPEND', 'TIKTOK_SPEND'
]

corr = df[cannibalization].corr()

print(corr)
```

### Negative correlation means potential cannibalization.
