### Google,Meta is the  most used Marketing source

```python
df.groupby('ORGANISATION_PRIMARY_TERRITORY_NAME')['ORGANISATION_ID'].nunique().reset_index()
```

```python
df['Total_spends']=df.loc[:,['GOOGLE_PAID_SEARCH_SPEND','GOOGLE_SHOPPING_SPEND','GOOGLE_PMAX_SPEND','GOOGLE_DISPLAY_SPEND','GOOGLE_VIDEO_SPEND','META_FACEBOOK_SPEND','META_INSTAGRAM_SPEND','META_OTHER_SPEND','TIKTOK_SPEND']].sum(axis=1)
```

```python
df.CURRENCY_CODE.unique()
```

```python
exchange_to_gbp={'USD':0.75,'GBP':1.0,'EUR':0.87,'AUD':0.49,'JPY':0.005,'HKD':0.097,'CAD':0.54,'CLP':0.00079,'ZAR':0.043,
       'NZD':0.43, 'SEK':0.079, 'DKK':0.12, 'NOK':0.074, 'CHF':0.94}

df['Total_spend_GBP']=df['Total_spends']*df['CURRENCY_CODE'].map(exchange_to_gbp)
```

```python
df['Total_Revenue']=df['ALL_PURCHASES_ORIGINAL_PRICE']-df['ALL_PURCHASES_GROSS_DISCOUNT']
```

```python
df.groupby('ORGANISATION_MARKETING_SOURCES')['Total_spend_GBP'].sum().plot(kind='bar')
```

### Google,Meta has the highest spend followed by Google,Meta,Tiktok

### Most spends are in Pounds so we have converted the spends currency to GBP

```python
df.Total_spend_GBP=df.Total_spend_GBP.astype('int64')
```

```python
df['Total_Revenue_GBP']=df['Total_Revenue']*df['CURRENCY_CODE'].map(exchange_to_gbp)
```

```python
df.Total_Revenue_GBP=df.Total_Revenue_GBP.astype('int64')
```

```python
df.groupby(['ORGANISATION_VERTICAL','ORGANISATION_SUBVERTICAL'])[['Total_spend_GBP','Total_Revenue_GBP']].sum().reset_index()
```

```python
df.groupby('ORGANISATION_MARKETING_SOURCES')['Total_Revenue_GBP'].sum().plot(kind='bar')
```

### Highest revenue is from Google, Meta

```python
seasonality=df.groupby(df['DATE_DAY'].dt.to_period('M'))[['Total_Revenue_GBP','Total_spend_GBP']].sum()
```
